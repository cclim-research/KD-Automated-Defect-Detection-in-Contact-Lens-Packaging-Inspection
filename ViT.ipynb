```python
import os
import torch
from torch.utils.data import Dataset, DataLoader
import numpy as np
from PIL import Image
from torchvision import transforms
from vit_pytorch import ViT
import torch.optim as optim
import torch.nn as nn
from pathlib import Path
from torchvision import datasets, models, transforms
from sklearn.model_selection import train_test_split
import shutil
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
```

```python
#Setup Path
FILTERED_DIR = r"D:\Adrian Leong\Coding\OneDrive_1_1-14-2025\Filtered Dataset"  # Replace with actual dataset path
AUGMENTED_DIR = os.path.join(FILTERED_DIR, "augmented")
SPLITS = ["train", "val", "test"]
RATIO = {"train": 0.8, "val": 0.1, "test": 0.1}
BATCH_SIZE = 32
DEVICE = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
```

```python
#Ensure necessary directories
for split in SPLITS:
    for label in ["PASS", "REJECT"]:
        os.makedirs(os.path.join(AUGMENTED_DIR, split, label), exist_ok=True)
```

```python


# Ensure necessary directories
for split in SPLITS:
    for label in ["PASS", "REJECT"]:
        os.makedirs(os.path.join(AUGMENTED_DIR, split, label), exist_ok=True)

# Safe hue adjustment function
def safe_hue_adjust(img, hue_factor):
    """ Adjust hue safely to prevent uint8 overflow. """
    img_hsv = img.convert("HSV")  # Convert to HSV
    np_hsv = np.array(img_hsv, dtype=np.int16)  # Use int16 to prevent overflow
    np_hsv[..., 0] = np.clip(np_hsv[..., 0] + int(hue_factor * 90), 0, 255)  # Scale hue safely
    return Image.fromarray(np_hsv.astype(np.uint8), "HSV").convert("RGB")  # Convert back to RGB

# Custom ColorJitter with safe hue adjustment
class SafeColorJitter:
    def __init__(self, brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1):
        self.color_jitter = transforms.ColorJitter(brightness, contrast, saturation, hue)

    def __call__(self, img):
        img = self.color_jitter(img)  # Apply standard ColorJitter
        return safe_hue_adjust(img, hue_factor=np.random.uniform(-0.1, 0.1))  # Random hue shift safely

# Updated transformations
transformations = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomAffine(degrees=30, translate=(0.1, 0.1), scale=(0.9, 1.1)),
    SafeColorJitter(),  # Use our safe jitter
    transforms.RandomPerspective(distortion_scale=0.5, p=0.5),
    transforms.Resize((224, 224))
])

# Apply transformations
for label in ["PASS", "REJECT"]:
    label_dir = os.path.join(FILTERED_DIR, label)
    augmented_label_dir = os.path.join(AUGMENTED_DIR, "all", label)
    os.makedirs(augmented_label_dir, exist_ok=True)

    for image_name in os.listdir(label_dir):
        image_path = os.path.join(label_dir, image_name)
        image = Image.open(image_path).convert("RGB")

        for i in range(8):  # Augment to get 8x data
            augmented_image = transformations(image)
            augmented_image.save(os.path.join(augmented_label_dir, f"{Path(image_name).stem}_aug{i}.jpg"))
```

```python
#Split Dataset
for label in ["PASS", "REJECT"]:
    augmented_label_dir = os.path.join(AUGMENTED_DIR, "all", label)
    images = [os.path.join(augmented_label_dir, img) for img in os.listdir(augmented_label_dir)]

    train, test_val = train_test_split(images, test_size=(1 - RATIO["train"]))
    val, test = train_test_split(test_val, test_size=(RATIO["test"] / (RATIO["val"] + RATIO["test"])))

    for split, split_images in zip(SPLITS, [train, val, test]):
        split_dir = os.path.join(AUGMENTED_DIR, split, label)
        for img_path in split_images:
            shutil.copy(img_path, split_dir)
```

```python
#Load Data and Label It
class CustomDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform
        self.image_paths = []
        self.labels = []

        for label, folder in enumerate(["PASS", "REJECT"]):
            folder_path = os.path.join(root_dir, folder)
            for img_name in os.listdir(folder_path):
                self.image_paths.append(os.path.join(folder_path, img_name))
                self.labels.append(label)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        image = Image.open(img_path).convert("RGB")
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label

#Prepare Data Loaders
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

train_dataset = CustomDataset(root_dir=os.path.join(AUGMENTED_DIR, "train"), transform=transform)
val_dataset = CustomDataset(root_dir=os.path.join(AUGMENTED_DIR, "val"), transform=transform)
test_dataset = CustomDataset(root_dir=os.path.join(AUGMENTED_DIR, "test"), transform=transform)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)
```

```python
# Define Vision Transformer Model
model = ViT(
    image_size=224,  # Size of input image
    patch_size=16,   # Patch size
    num_classes=2,   # Number of classes (PASS, REJECT)
    dim=768,         # Dimension of transformer
    depth=12,        # Depth of transformer
    heads=12,        # Number of attention heads
    mlp_dim=3072,    # Size of the feed-forward layer
    dropout=0.1,     # Dropout rate
    emb_dropout=0.1  # Dropout rate for the embedding layer
)

# Move model to device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Define Optimizer and Loss Function
optimizer = optim.Adam(model.parameters(), lr=1e-4)
criterion = nn.CrossEntropyLoss()

# Create lists to store training and validation losses/accuracies
train_losses = []
train_accuracies = []
val_losses = []
val_accuracies = []
```

```python
# Training Loop
num_epochs = 10
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        
        # Calculate accuracy for training set
        _, preds = torch.max(outputs, 1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)

    train_losses.append(running_loss / len(train_loader))
    train_accuracies.append(correct / total)

    # Validation Phase
    model.eval()
    val_loss = 0.0
    correct = 0
    total = 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            val_loss += loss.item()
            
            _, preds = torch.max(outputs, 1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)

    val_losses.append(val_loss / len(val_loader))
    val_accuracies.append(correct / total)

    print(f"Epoch [{epoch+1}/{num_epochs}], "
          f"Train Loss: {train_losses[-1]:.4f}, Train Accuracy: {train_accuracies[-1]:.4f}, "
          f"Val Loss: {val_losses[-1]:.4f}, Val Accuracy: {val_accuracies[-1]:.4f}")

# After training, plot the metrics
def plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies, num_epochs):
    epochs = range(num_epochs)

    # Plot Training and Validation Loss
    plt.figure(figsize=(10, 5))
    plt.plot(epochs, train_losses, label='Training Loss')
    plt.plot(epochs, val_losses, label='Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.show()

    # Plot Training and Validation Accuracy
    plt.figure(figsize=(10, 5))
    plt.plot(epochs, train_accuracies, label='Training Accuracy')
    plt.plot(epochs, val_accuracies, label='Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.title('Training and Validation Accuracy')
    plt.legend()
    plt.show()

# Plot the metrics after training
plot_metrics(train_losses, val_losses, train_accuracies, val_accuracies, num_epochs)
```

```python
# Evaluate model
model.eval()
y_true = []
y_pred = []

with torch.no_grad():
    for inputs, labels in test_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = model(inputs)
        _, preds = torch.max(outputs, 1)
        y_true.extend(labels.cpu().numpy())
        y_pred.extend(preds.cpu().numpy())

# Calculate Metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average='weighted')
recall = recall_score(y_true, y_pred, average='weighted')
f1 = f1_score(y_true, y_pred, average='weighted')

print(f"Test Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
```
